```python
name= input("Enter your name: ")

weight = int(input("Enter your weight in pounds:"))

height = int(input("Enter your height in inches:"))

BMI = (weight *703) / (height *height )

print (BMI)

if BMI > 0:
    if BMI < 18.5:
        print(name + ", You are underweight.")
    elif BMI <= 25:
        print(name + ", You are normal weight.")
    elif BMI <= 30:
        print(name + ", You are overweight. You need to exercies more and stop setting and writing so many python code")
    elif BMI <= 35:
        print(name + ", You are Obese Class I.")
    elif BMI <= 40:
        print(name + ", You are Obese Class II.")
    else:
        print(name + ", You are Obese Class III.")
else:
    print ("Enter valid input")
```
